```python
import pandas as pd

DATA_URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/magic/magic04.data"

COLUMNS = ["fLength","fWidth","fSize","fConc",
           "fConc1","fAsym","fM3Long","fM3Trans",
           "fAlpha","fDist","class"]

df = pd.read_csv(DATA_URL, header=None, names=COLUMNS)
print(df.shape)
df.head()
```

```python
#label encoding
df["y"]=(df["class"]=="g").astype(int)
print(df["y"].value_counts())
```

```python
# train/test 분리 (8:2)
from sklearn.model_selection import train_test_split
train_df, test_df = train_test_split(
    df, test_size=0.2, stratify=df["y"], random_state=42
)
```

```python
# 9. 피처/타겟 분리
NUM_COLS = ["fLength","fWidth","fSize","fConc","fConc1",
            "fAsym","fM3Long","fM3Trans","fAlpha","fDist"]
# 숫자 피처 10개 목록

X_train = train_df[NUM_COLS].to_numpy()
# train 데이터에서 피처 10개만 뽑아서 numpy 배열로
y_train = train_df["y"].to_numpy()
# train 데이터에서 Label(0/1)만 뽑아서 numpy 배열로

X_test = test_df[NUM_COLS].to_numpy()
y_test = test_df["y"].to_numpy()
# test도 똑같이
```

```python
# 10. 스케일링 (선형모델용)
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(X_train)          # 여기서의 fit는 평균,표준편차를 계산해서 scaler에 저장
X_train_lin = scaler.transform(X_train)  # 계산한 평균 표준편차로 훈련 데이터 변환
X_test_lin = scaler.transform(X_test)    # 테스트 데이터 변환
# fit은 훈련만, transform은 둘 다
# → leakage 방지

# 트리 모델용은 스케일링 불필요
X_train_tree = X_train
X_test_tree = X_test
```

```python
# ==========================================
# 1. 로지스틱 회귀 (Logistic Regression) 학습 및 평가
# ==========================================
# 선형 모델이므로 표준화 스케일링이 완료된 'X_train_lin', 'X_test_lin'을 사용합니다.

from sklearn.linear_model import LogisticRegression

# 모델 객체 생성 및 하이퍼파라미터 설정
lr_model = LogisticRegression(
    max_iter=1000,       # 최적화 알고리즘이 반복할 최대 횟수
    C=1.0,               # 규제 강도의 역수 (작을수록 강한 규제 -> 과적합 방지)
    solver="lbfgs",      # 최적화에 사용할 알고리즘 종류
    random_state=42      # 앞서 사용하신 random_state=42와 통일하여 재현성 보장
)

# 모델 학습 (스케일링된 훈련 데이터 사용)
lr_model.fit(X_train_lin, y_train)

# 테스트 데이터 예측 (클래스 0 또는 1)
y_pred_lr = lr_model.predict(X_test_lin)

# 테스트 데이터가 각 클래스에 속할 확률 예측
y_score_lr = lr_model.predict_proba(X_test_lin)


# ==========================================
# 2. 의사결정나무 (Decision Tree) 학습 및 평가
# ==========================================
# 트리 기반 모델은 데이터의 절대적 크기에 영향을 받지 않으므로 원본 'X_train_tree', 'X_test_tree'를 사용합니다.

from sklearn.tree import DecisionTreeClassifier

# 모델 객체 생성 및 하이퍼파라미터 설정
dt_model = DecisionTreeClassifier(
    max_depth=8,            # 트리의 최대 깊이 제한 (과적합 방지)
    min_samples_leaf=20,    # 말단 노드가 되기 위한 최소 샘플 수 (과적합 방지)
    random_state=42         # 난수 고정
)

# 모델 학습 (스케일링되지 않은 원본 데이터 사용)
dt_model.fit(X_train_tree, y_train)

# 테스트 데이터 예측 (클래스 0 또는 1)
y_pred_dt = dt_model.predict(X_test_tree)

# 테스트 데이터가 '1(g, gamma)' 클래스에 속할 확률만 추출
# [:, 1]은 [0일 확률, 1일 확률] 중 두 번째 열(인덱스1=>숫자 1)만 가져옵니다. :은 해당열의 모든 행을 의미합니다.
y_score_dt = dt_model.predict_proba(X_test_tree)[:, 1]


# ==========================================
# 3. 모델 예측 결과 확인 (간단한 출력 예시)
# ==========================================
# 이미지에 있던 evaluate 함수 대신 사이킷런 기본 평가지표로 결과를 출력해 봅니다.
#accuracy_score, classification_report 등을 사용하여 모델 성능을 평가합니다. 스스로 채점을 시킨다
from sklearn.metrics import accuracy_score, classification_report

print("=== [Logistic Regression] 결과 ===")
print(f"정확도(Accuracy): {accuracy_score(y_test, y_pred_lr):.4f}")
print(f"첫 5개 예측 확률(1일 확률): {y_score_lr[:5, 1]}\n")

print("=== [Decision Tree] 결과 ===")
print(f"정확도(Accuracy): {accuracy_score(y_test, y_pred_dt):.4f}")
print(f"첫 5개 예측 확률(1일 확률): {y_score_dt[:5]}")
```
