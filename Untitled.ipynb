```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

# train 불러오기 및 전처리
df = pd.read_csv("train.csv")
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
df["Embarked"] = df["Embarked"].map({"S": 0, "C": 1, "Q": 2})
df["Fare"] = df["Fare"].fillna(df["Fare"].median())

X = df[["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]]
y = df["Survived"]

# test 불러오기 및 전처리
test = pd.read_csv("test.csv")
test["Age"] = test["Age"].fillna(test["Age"].median())
test["Sex"] = test["Sex"].map({"male": 0, "female": 1})
test["Embarked"] = test["Embarked"].map({"S": 0, "C": 1, "Q": 2})
test["Fare"] = test["Fare"].fillna(test["Fare"].median())

X_test = test[["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]]

# 모델 학습 및 예측
model = RandomForestClassifier(n_estimators=100, max_depth=8, min_samples_split=5, random_state=42)
model.fit(X, y)
predictions = model.predict(X_test)

# 제출 파일 저장
submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": predictions
})
submission.to_csv("submission4.csv", index=False)
print("저장 완료!")
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

# train 불러오기 및 전처리
df = pd.read_csv("train.csv")
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
df["Embarked"] = df["Embarked"].map({"S": 0, "C": 1, "Q": 2})
df["Fare"] = df["Fare"].fillna(df["Fare"].median())
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

# test 불러오기 및 전처리
test = pd.read_csv("test.csv")
test["Age"] = test["Age"].fillna(test["Age"].median())
test["Sex"] = test["Sex"].map({"male": 0, "female": 1})
test["Embarked"] = test["Embarked"].map({"S": 0, "C": 1, "Q": 2})
test["Fare"] = test["Fare"].fillna(test["Fare"].median())
test["FamilySize"] = test["SibSp"] + test["Parch"] + 1
test["IsAlone"] = (test["FamilySize"] == 1).astype(int)

X = df[["Pclass", "Sex", "Age", "Fare", "Embarked", "FamilySize", "IsAlone"]]
y = df["Survived"]
X_test = test[["Pclass", "Sex", "Age", "Fare", "Embarked", "FamilySize", "IsAlone"]]

# 하이퍼파라미터 튜닝
params = {
    "n_estimators": [100, 200, 300],
    "max_depth": [5, 6, 7, 8],
    "min_samples_split": [5, 10, 15]
}

grid = GridSearchCV(RandomForestClassifier(random_state=42), params, cv=5)
grid.fit(X, y)

print("최적 파라미터:", grid.best_params_)
print("최고 점수:", grid.best_score_)

# 제출 파일 저장
predictions = grid.predict(X_test)
submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": predictions
})
submission.to_csv("submission9.csv", index=False)
print("저장 완료!")
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score

df = pd.read_csv("train.csv")
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})
df["Embarked"] = df["Embarked"].map({"S": 0, "C": 1, "Q": 2})
df["Fare"] = df["Fare"].fillna(df["Fare"].median())
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)
df["Title"] = df["Name"].str.extract(r',\s*([^\.]+)\.')
df["Title"] = df["Title"].map({"Mr": 0, "Miss": 1, "Mrs": 2, "Master": 3}).fillna(4)

test = pd.read_csv("test.csv")
test["Age"] = test["Age"].fillna(test["Age"].median())
test["Sex"] = test["Sex"].map({"male": 0, "female": 1})
test["Embarked"] = test["Embarked"].map({"S": 0, "C": 1, "Q": 2})
test["Fare"] = test["Fare"].fillna(test["Fare"].median())
test["FamilySize"] = test["SibSp"] + test["Parch"] + 1
test["IsAlone"] = (test["FamilySize"] == 1).astype(int)
test["Title"] = test["Name"].str.extract(r',\s*([^\.]+)\.')
test["Title"] = test["Title"].map({"Mr": 0, "Miss": 1, "Mrs": 2, "Master": 3}).fillna(4)

X = df[["Pclass", "Sex", "Age", "Fare", "Embarked", "FamilySize", "IsAlone", "Title"]]
y = df["Survived"]
X_test = test[["Pclass", "Sex", "Age", "Fare", "Embarked", "FamilySize", "IsAlone", "Title"]]

model = RandomForestClassifier(n_estimators=200, max_depth=7, min_samples_split=10, random_state=42)
scores = cross_val_score(model, X, y, cv=5)
print("교차검증 평균:", scores.mean())

model.fit(X, y)
predictions = model.predict(X_test)

submission = pd.DataFrame({
    "PassengerId": test["PassengerId"],
    "Survived": predictions
})
submission.to_csv("submission10.csv", index=False)
print("저장 완료!")
```
